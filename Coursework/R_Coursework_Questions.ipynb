### Question 7

The `dplyr` dataset `storms` is part of the US National Oceanic and Atmospheric Administration database of Atlantic hurricanes. It contains multiple repeated measurements from hurricanes. For a given storm, there are measurements from every 6 hours across the life time of the storm.

Use a tidyverse approach with `dplyr` to answer the following:

- how many different storms are contained in the database?
- which year had the most storms?
- which storm, in which year had the highest pressure?

HINT:
unique() will help ensure you remove duplicated data from a vector e.g.
x <- c(1, 1, 2, 2, 3, 3, 4, 4)
unique(x)

```r
#install dplyr package if not already installed
#install.packages("dplyr")

#bringing dplyr package into script
library(dplyr) 

#loading storms data set in
data(storms)

#finding how many different storms are in the database
storms %>%                #using pipes to help simplify the data manipulation
    distinct(name) %>%    #distinct function used to see unique entries
    tally()               #prints the number of how many different entries

#finding which year had the most storms
storms %>%
    distinct(year, name) %>% #using distinct to see unique entries for each year
    count(year) %>%          #counts how many individual storms in each year
    arrange(desc(n)) %>%     #arranges in order from most to least
    head(1)                  #prints the year with most storms & n/o storms

#finding out which storm had the highest pressure and the year
storms %>%
    arrange(desc(pressure)) %>%
    select(name, year, pressure) %>%  #selects just the specified columns
    head(1)
```

### Question 8

The `ggplot2` dataset `diamonds` contains the price and information from the sale of 54,000 different diamonds. You can use ?diamonds to find out more about the dataset if you want.

Using `ggplot2`:

- plot the relationship between the carat (i.e. the weight of the diamond) and the price.
- plot a histogram of prices of all the diamonds
- plot a boxplot of the difference in price between the different cuts

```r
#install ggplot2 package if not already installed
#install.packages("ggplot2")

#bringing ggplot2 package into script
library(ggplot2) 

#loading diamonds data set in
data(diamonds)

#plotting the relationship between the carat and the price
ggplot(aes(x=carat, y=price), data = diamonds) +
    geom_point() +
    labs(     #using labs function to add title
        title = "Plot to show the relationship between the carat and price of diamonds"
        )

#plotting histogram of the price of all the diamonds
ggplot(aes(x=price), data = diamonds) + 
    geom_histogram(binwidth = 500, fill = "palegreen1", colour = "palegreen4") +
     labs(     #using labs function to add title
        title = "Plot to show the relationship between the carat and price of diamonds"
         )
        
#plotting boxplot of difference in price between diffserent cuts
ggplot(aes(x= cut, y=price), data = diamonds)+
    geom_boxplot(fill = "deepskyblue", colour = "dodgerblue")+
  labs(     #using labs function to add title
    title = "Plot to show the difference between in price between different cuts"
  ) 
```

### Question 9

Identify the maximum and minimum price that the diamonds in the diamonds data set sold for. Add a new column to the diamonds data set that contains the string “Low price” if the value in the price column is less than 5000 dollars and “High price” if it is above 5000 dollars.

```r
#install dplyr package if not already installed
#install.packages("dplyr")

#install ggplot2 package if not already installed
#install.packages("ggplot2")

#bringing dplyr package into script
library(dplyr)

#bringing ggplot2 package into script
library(ggplot2)

#loading diamonds data set in
data(diamonds)

#identifying min and max price diamonds are sold for
diamonds %>%
    summarise(                                   #produces a tibble with the minimum
        min_price = min(price, na.rm = TRUE),    #price and mamximum displayed
        max_price = max(price, na.rm = TRUE) 
        )

#adding in new column based on price
diamonds <- diamonds %>%             #adds new column to existing data set
    mutate(price_category = ifelse(price <5000, "Low Price", "High Price")) 
head(diamonds)            
```

### Question 10

Sticking with the `diamonds` dataset for this question. A carat is a special unit of weight for gemstones where a single carat is 200 mg or 0.2 g. Write a function to convert the `diamonds$carat` vector into grams or milligrams. How heavy is the heaviest diamond?

```r
#install ggplot2 package if not already installed
#install.packages("ggplot2")

#bringing ggplot2 package into script
library(ggplot2)

#loading diamonds data set in
data(diamonds)

#function to convert carats to grams
convert <- function(x) {
    return(x * 0.2) 
    }

#creates new vector with the converted weights
converted_weight <- convert(diamonds$carat)    

#to find heaviest diamond
heaviest <- max(converted_weight) 
print(paste("The heaviest diamond weighs", heaviest, "grams."))   #prints the weight, using paste to join the text and output 
```
