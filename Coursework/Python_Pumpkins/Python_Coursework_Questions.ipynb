### Question 3

Use three different methods to print the numbers 200 to 300.

````python
#using for loops
print("Using for loops to print the numbers 200 to 300")   #adding the text in between to help for easier viewing
for x in range(200,301):
    print(x) 

#using while loop
print("Using while loops to print the numbers 200 to 300") 
i = 200
while i < 301:
    print(i)
    i = i + 1

#using a list
print("Using a list to print the numbers 200 to 300")
to_print = list(range(200,301))
print(to_print)
````

### Question 6
Using the startwars dataset imported below, identify the individuals with tallest height and lowest mass.

````python
starwars_dataset="https://www.fabricionarcizo.com/post/starwars/updated_starwars.csv"
starwars = pd.read_csv(starwars_dataset)
````

````python
#importing libraries
import pandas as pd

#finding tallest individual
tallest = starwars.sort_values("height", ascending=False).iloc[[0]] #using iloc to give location of sorted list not original list
print(tallest)

#finding lightest individual
lightest = starwars.sort_values("mass").iloc[[0]]
print(lightest)
````

### Question 7
Analyze the life expectancy dataset (which can be loaded by running the cell below).

- Calculate the number of unique countries in the dataset.

- Calculate the country and year with the highest life expectancy.

- Calculate the year with the highest total expenditure.

- Provide your code using pandas.


````python
import seaborn as sns
life_expectancy = sns.load_dataset('healthexp')
````

````python
print(life_expectancy.columns) #looking at column names for unknown data set
````

````python
#calculating number of unique countries
no_countries = life_expectancy["Country"].nunique()
print("The number of unique countries is", no_countries)

#country and year with highest life expectancy
life_expect = life_expectancy.sort_values("Life_Expectancy",  ascending=False).iloc[0]
country = life_expect["Country"]
year = life_expect["Year"]
print("The country with the highest life expectancy is", country, "in", year)

#year with highest total expenditure
expend = life_expectancy.sort_values("Spending_USD",  ascending=False).iloc[0]
country2 = expend["Country"]
cost = expend["Spending_USD"]
print("The country with the highest total expenditure is", country2, "with $", cost) 
````

Using the diamonds dataset and the plotting code of your choice, plot the following from the diamonds dataset.

````python
#loading libraries
import seaborn as sns
import matplotlib.pyplot as plt 

#scatter plot of carat v price
sns.scatterplot(data=diamonds, x="carat", y="price")
plt.title("Scatter to show Carat v Price")
plt.show()

#histogram of prices
sns.histplot(data=diamonds, x="price", bins=30)
plt.title("Histogram of Prices")
plt.show()

#boxplot of cut v price
sns.boxplot(data=diamonds, x="cut", y="price")
plt.title("Boxplot to show Price by Cut")
plt.show()
````

Using the same dataset, identify the max and min price in diamonds dataset.

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.
1 carat = 0.2 grams

````python
#loading libraries
import seaborn as sns

#loading data
diamonds = sns.load_dataset('diamonds')

#defining function
def convert(carat):
    return carat * 0.2

#converting 
conversion = convert(diamonds["carat"])

#finding heaviest diamond
heaviest = conversion.max()
print("The heaviest diamond weighs", heaviest, "grams")
````
